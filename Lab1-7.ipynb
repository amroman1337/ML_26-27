Набор данных для изучения зависимости различных параметров тела человека.

Файл: `data.csv`

```

Cognitive Function - самооценка когнитивных функций по шкале от 0 (плохая) до 100 (отличная).

Education Level - уровень образования.
Income Level - доход в USD.
```

````python
import pandas as pd
````

````python
df = pd.read_csv('data.csv', keep_default_na=False)
print('Размер таблицы:', df.shape)
print(df.head())

men = df['Gender'] == 'Male'
women = df['Gender'] == 'Female'
income_group = df['Income Level']
income_low = income_group == 'Low'
income_high = income_group == 'High'
sun_high = df['Sun Exposure'] > df['Sun Exposure'].quantile(0.75)
diseases = df['Chronic Diseases']
no_disease = diseases == 'None'
has_disease = (~no_disease).astype(int)
````

1. Какой средний вес людей представленных в наборе данных?

````python
print('Средний вес:', round(df['Weight (kg)'].mean(), 2), 'кг')
````

2. Сбалансировано ли соотношение мужчин и женщин в наборе данных?

````python
counts = df['Gender'].str.strip().str.lower().value_counts()
print(counts)
print('Доли, %:')
print((counts / counts.sum() * 100).round(2))
male_share = men.sum() / (men.sum() + women.sum())
print('Соотношение примерно сбалансировано:', 0.47 <= male_share <= 0.53)
````

3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
bp = df['Blood Pressure (s/d)'].str.split('/', expand=True).astype(float)
bp.columns = ['Систолическое', 'Диастолическое']
low_stress = df['Stress Levels'] <= df['Stress Levels'].quantile(0.25)
high_stress = df['Stress Levels'] >= df['Stress Levels'].quantile(0.75)
pressure = pd.DataFrame({'Низкий стресс': bp[low_stress].mean(), 'Высокий стресс': bp[high_stress].mean()})
print(pressure)
print('При высоком стрессе среднее давление выше:')
print(pressure['Высокий стресс'] > pressure['Низкий стресс'])
````

4. Правда ли, что уровень холестерина у мужчин и женщин отличается?

````python
cholesterol = df.groupby('Gender')['Cholesterol Level (mg/dL)'].agg(['count', 'mean'])
print(cholesterol)
print('Разница средних (мужчины минус женщины):',
      df.loc[men, 'Cholesterol Level (mg/dL)'].mean()
      - df.loc[women, 'Cholesterol Level (mg/dL)'].mean())
````

5. Правда ли, что с возрастом женщины имеют свойство оценивать свои когнитивные способности хуже, чем в юности (порог юности - 30 лет, женщина в возрасте - 60 лет+)

````python
young = df.loc[women & (df['Age (years)'] < 30), 'Cognitive Function']
older = df.loc[women & (df['Age (years)'] >= 60), 'Cognitive Function']
print('До 30 лет: число наблюдений =', young.count(), ', среднее =', young.mean())
print('От 60 лет: число наблюдений =', older.count(), ', среднее =', older.mean())
if young.count() > 0 and older.count() > 0:
    print('В старшей группе средняя оценка ниже:', older.mean() < young.mean())
else:
    print('Для сравнения недостаточно данных')
````

````python
print('Минимальный вес мужчин с высоким заработком:', df.loc[men & income_high, 'Weight (kg)'].min())
print('Минимальный вес мужчин с низким заработком:', df.loc[men & income_low, 'Weight (kg)'].min())
````

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
q30 = df['BMI'].quantile(0.30)
q60 = df['BMI'].quantile(0.60)
q75 = df['BMI'].quantile(0.75)
bmi_high = df['BMI'] > q75
bmi_middle = df['BMI'].between(q30, q60)
print('Число известных ответов:', has_disease[bmi_high].count(), has_disease[bmi_middle].count())
share_high = has_disease[bmi_high].mean()
share_middle = has_disease[bmi_middle].mean()
print('Доля заболеваний при BMI выше 75-го перцентиля:', share_high)
print('Доля заболеваний при BMI между 30-м и 60-м перцентилями:', share_middle)
if pd.notna(share_high) and pd.notna(share_middle):
    print('В первой группе заболевания встречаются чаще:', share_high > share_middle)
````

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
activity = df['Physical Activity Level'].str.strip().str.title()
print(pd.crosstab(income_group, activity, normalize='index').round(3))
activity_score = activity.map({'Low': 1, 'Moderate': 2, 'Medium': 2, 'High': 3})
low_mean = activity_score[income_low].mean()
high_mean = activity_score[income_high].mean()

print('Средний балл активности при низком доходе:', low_mean)
print('Средний балл активности при высоком доходе:', high_mean)
if pd.notna(low_mean) and pd.notna(high_mean):
    print('При высоком доходе активность выше:', high_mean > low_mean)
````

10. Какая доля мужчин с высокой частотой пребывания на солнце имеет высокий заработок?

````python
selected = men & sun_high
known = selected & income_group.notna()

if known.sum() > 0:
    share = income_high[known].mean()
    print('Доля мужчин с высоким заработком:', round(share * 100, 2), '%')
else:
    print('Нет подходящих наблюдений с известным заработком')
````

5. Постройте круговую диаграмму отображающую как ментальное здоровье связано с паттернами сна

````python
table = pd.crosstab(df['Sleep Patterns'], df['Mental Health Status'])
colors = {'Poor': 'salmon', 'Fair': 'gold', 'Good': 'lightgreen', 'Excellent': 'skyblue'}
fig, axes = plt.subplots(1, len(table), figsize=(15, 5))
for ax, (pattern, row) in zip(axes, table.iterrows()):
    row.plot.pie(ax=ax, autopct='%1.1f%%', colors=[colors[x] for x in row.index])
    ax.set_title('Сон: ' + pattern)
    ax.set_ylabel('')
plt.suptitle('Ментальное здоровье при разных паттернах сна')
plt.tight_layout()
plt.show()
print(pd.crosstab(df['Sleep Patterns'], df['Mental Health Status'], normalize='index').round(3))
````
